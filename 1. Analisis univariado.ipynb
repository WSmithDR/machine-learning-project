```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from constants import initial_dir 
```

```python
df_test = pd.read_csv(f'{initial_dir}/vehiculos_test.csv')
df_train = pd.read_csv(f'{initial_dir}/vehiculos_train.csv')
```

## Análisis Univariado:
- Seleccione al menos tres variables y realice un análisis univariable en cada una.
- Proporcione visualizaciones (por ejemplo, histogramas, diagramas de caja) para cada variable seleccionada e interprételas.

### Variable 1: Precio del Vehiculo

```python
variable_analisis = 'precio_vehiculo'
```

#### df_train

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_train[variable_analisis], kde=True)
plt.title(f'Distribución de {variable_analisis} del dataset train')
plt.xlabel(f'{variable_analisis}')
plt.ylabel('Frecuencia')
plt.show()
```

```python
df_train[[variable_analisis]].boxplot(figsize=(10, 10))
plt.tight_layout()
plt.title(f'Diagrama de caja del {variable_analisis} del dataset train')
plt.xticks(rotation=45, ha='right')
plt.show()
```

Análisis: Se observan datos aberrantes, valores de precio en negativo y valores atípicos.

#### df_test

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_test[variable_analisis], kde=True)
plt.title(f'Distribución de {variable_analisis} del dataset test')
plt.xlabel(f'{variable_analisis}')
plt.ylabel('Frecuencia')
plt.show()
```
