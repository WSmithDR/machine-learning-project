```python
df_train_numeric = df_train.select_dtypes(include=['number'])
df_test_numeric = df_test.select_dtypes(include=['number'])
df_train_numeric=df_train_numeric.loc[(df_train_numeric["anio_registro"]<2025)&(df_train_numeric['anio_registro']>=1995)&(df_train_numeric['millas_recorridas']<250000)]
df_test_numeric=df_test_numeric.loc[(df_test_numeric["anio_registro"]<2025)&(df_test_numeric['anio_registro']>=1995)&(df_test_numeric['millas_recorridas']<250000)]
```

La mayoria de los casos de fraudes están registrados en las millas recorridas negativas por lo que eliminarlos no son una opción. Vemos que en los vehiculos cero millas, solo se ha registrado un caso de fraude, el resto presumiblemente fueron vehiculos que llegaron defectos de frábica, por lo que el cero no se corresponde a un dato atipico. Parecer ser que la mejor solucion es simplemente convertir a positivos las millas negativas, valores que estarían dentro de los rangos normales.

```python
mask = df_train['millas_recorridas'] < 0
```

```python
df_train.loc[mask, 'millas_recorridas'] = df_train.loc[mask, 'millas_recorridas'].abs()
```

```python
df_test.loc[mask, 'millas_recorridas'] = df_test.loc[mask, 'millas_recorridas'].abs()
```
