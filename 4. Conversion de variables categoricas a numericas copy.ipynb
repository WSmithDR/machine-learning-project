```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from constants import tratados_dir
```

```python
df_test = pd.read_csv(f'{tratados_dir}/vehiculos_test.csv')
df_train = pd.read_csv(f'{tratados_dir}/vehiculos_train.csv')
```

Definir el umbral mínimo de frecuencia para considerar una marca como representativa utlizando la diferencia entre la media y la desviaicon estandar.

```python
umbral = 6000
```

```python


 # por ejemplo, marcas con menos de 50 apariciones se agrupan en "OTROS"

# Calcular la frecuencia de cada marca
frecuencias = df_train['marca'].value_counts()
```

```python
frecuencias
```

```python
# Create new column 'marca2' where rare brands are renamed to 'OTROS'
df_train['marca2'] = df_train['marca'].apply(lambda x: x if frecuencias[x] >= umbral else 'OTROS')

df_test['marca2'] = df_test['marca'].apply(lambda x: x if frecuencias[x] >= umbral else 'OTROS')
```

```python
# Check the new distribution of brands
print("Train: ", df_train['marca2'].value_counts())
print("Test: ", df_test['marca2'].value_counts())
```

```python
# Aplicamos one-hot encoding a la columna 'marca2'
df_train_encoded = pd.get_dummies(df_train, columns=['marca2'])

# Si deseas que los valores codificados sean 0 y 1 (en número, no booleanos):
columnas_marca = [col for col in df_train_encoded.columns if col.startswith('marca2_')]

# Convertimos a numérico (0 y 1 en lugar de True/False)
for col in columnas_marca:
    df_train_encoded[col] = df_train_encoded[col].replace({True: 1, False: 0})
    df_train_encoded[col] = pd.to_numeric(df_train_encoded[col])
```

```python
# Aplicamos one-hot encoding a la columna 'marca2'
df_test_encoded = pd.get_dummies(df_test, columns=['marca2'])

# Si deseas que los valores codificados sean 0 y 1 (en número, no booleanos):
columnas_marca = [col for col in df_test_encoded.columns if col.startswith('marca2_')]

# Convertimos a numérico (0 y 1 en lugar de True/False)
for col in columnas_marca:
    df_test_encoded[col] = df_test_encoded[col].replace({True: 1, False: 0})
    df_test_encoded[col] = pd.to_numeric(df_test_encoded[col])

```
