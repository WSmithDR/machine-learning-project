## Separacion de los datasets

```python
import pandas as pd
from constants import predictor_vars, target_var, converted_dir
from sklearn.model_selection import train_test_split
```

```python
df_test = pd.read_csv(f"{converted_dir}/vehiculos_test.csv")
df_train = pd.read_csv(f"{converted_dir}/vehiculos_train.csv")
```

```python
X_train_val = pd.DataFrame(df_train[predictor_vars])
y_train_val = pd.DataFrame(df_train[target_var])
X_test = pd.DataFrame(df_test[predictor_vars])

```

```python
X_train, X_val, y_train, y_val = train_test_split(X_train_val, y_train_val, test_size=0.20, random_state=42)
```

Nos aseguramos que la reparticion entrenamiento-validacion esté conrrecta.

```python
print(f"Train: {len(X_train)/len(X_train_val)*100}%")
print(f"Validation: {len(X_val)/len(X_train_val)*100}%")
```

Procedemos a guardar los datos separados para implemntarlos en un próximo apartado.

```python
import os
from constants import separated_dir
```

```python
os.makedirs(separated_dir, exist_ok=True)
```

```python
X_train_ = pd.DataFrame(X_train, columns=predictor_vars)
X_train_.to_csv(f"{separated_dir}/X_train.csv", index=False)
```

```python
X_val_ = pd.DataFrame(X_val, columns=predictor_vars)
X_val_.to_csv(f"{separated_dir}/X_val.csv", index=False)
```

```python
y_train_ = pd.DataFrame(y_train, columns=[target_var])
y_train_.to_csv(f"{separated_dir}/y_train.csv", index=False)
```

```python
y_val_ = pd.DataFrame(y_val, columns=[target_var])
y_val_.to_csv(f"{separated_dir}/y_val.csv", index=False)
```

```python
X_test_ = pd.DataFrame(X_test, columns=predictor_vars)
X_test_.to_csv(f"{separated_dir}/X_test.csv", index=False)
```
