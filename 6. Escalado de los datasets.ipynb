## Escalado de los datasets

En este apartado hemos decidido realizar la normalizacion de datos debido a las diferentes magnitudes o rangos que manejan nuestras variables predictoras lo que podría traernos problemas a la hora de entrenar nuestros modelos ML.

```python
import pandas as pd
from constants import  predictor_vars, target_var, converted_dir
from sklearn.preprocessing import MinMaxScaler
from constants import separated_dir
```

```python
X_train = pd.read_csv(f"{separated_dir}/X_train.csv")
X_val = pd.read_csv(f"{separated_dir}/X_val.csv")
y_train = pd.read_csv(f"{separated_dir}/y_train.csv")
y_val = pd.read_csv(f"{separated_dir}/y_val.csv")
X_test = pd.read_csv(f"{separated_dir}/X_test.csv")
```

Contruimos y afinamos los escaladores para cada conjunto de variables.

```python
X_scaler = MinMaxScaler()
y_scaler = MinMaxScaler()
```

```python
X_scaler.fit(X_train)
y_scaler.fit(y_train)
```

Realizamos las transformaciones

```python
X_train_scaled = X_scaler.transform(X_train)
X_val_scaled = X_scaler.transform(X_val)
X_test_scaled = X_scaler.transform(X_test)
```

```python
y_train_scaled = y_scaler.transform(y_train)
y_val_scaled = y_scaler.transform(y_val)
```

Procedemos a guardar los datos escalados para implemntarlos en un próximo apartado.

```python
import os
from constants import scaled_dir
```

```python
os.makedirs(scaled_dir, exist_ok=True)
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled, columns=predictor_vars)
X_train_scaled.to_csv(f"{scaled_dir}/X_train.csv", index=False)
```

```python
X_val_scaled = pd.DataFrame(X_val_scaled, columns=predictor_vars)
X_val_scaled.to_csv(f"{scaled_dir}/X_val.csv", index=False)
```

```python
X_test_scaled = pd.DataFrame(X_test_scaled, columns=predictor_vars)
X_test_scaled.to_csv(f"{scaled_dir}/X_test.csv", index=False)
```

```python
y_train_scaled = pd.DataFrame(y_train_scaled, columns=[target_var])
y_train_scaled.to_csv(f"{scaled_dir}/y_train.csv", index=False)
```

```python
y_val_scaled = pd.DataFrame(y_val_scaled, columns=[target_var])
y_val_scaled.to_csv(f"{scaled_dir}/y_val.csv", index=False)
```
