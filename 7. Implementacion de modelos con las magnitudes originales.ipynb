```python
import pandas as pd
from constants import  separated_dir
```

### 2. Creacion de los dataframes

```python
X_val = pd.read_csv(f"{separated_dir}/X_val.csv")
X_train = pd.read_csv(f"{separated_dir}/X_train.csv")
y_val = pd.read_csv(f"{separated_dir}/y_val.csv")
y_train = pd.read_csv(f"{separated_dir}/y_train.csv")
```

### 4. Configuración de los Modelos
Implementamos tres variantes del Random Forest con diferentes enfoques:

### Modelo 1: Random Forest con Configuración Básica

```python
from sklearn.ensemble import RandomForestClassifier
```

Empezamos con el modelo mas sencillo donde practicamente usamos los valores por defecto de los parámetros

```python
rf1 = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42
)
```

```python
rf1.fit(X_train, y_train)
```

### Modelo 2: Random Forest con Optimización de Características

En este modelo elegimos seguir ciertas recomendaciones basadas en buenas prácticas y literatura sobre Random Forest. Se recomienda que el número de estimadores (`n_estimators`) esté entre 100 y 300 para lograr un buen equilibrio entre rendimiento y tiempo de cómputo; por ello, seleccionamos el punto medio de 200. Además, para evitar el sobreajuste y mejorar la capacidad de generalización del modelo, se sugiere limitar la profundidad máxima (`max_depth`) de los árboles. En este caso, aunque la recomendación es no superar una profundidad de 13, optamos por un valor conservador de 10.

También ajustamos otros hiperparámetros importantes: 
- `min_samples_split` se estableció en 5 para asegurar que cada división interna de los árboles tenga suficientes muestras, lo que ayuda a reducir el sobreajuste.
- `min_samples_leaf` se fijó en 2 para que cada hoja tenga al menos dos muestras, aportando mayor robustez.
- `class_weight` se configuró como `'balanced'` para compensar el desbalance en las clases del conjunto de datos, mejorando así la capacidad del modelo para identificar correctamente ambas clases.

```python
rf2 = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=2,
    class_weight='balanced',
    random_state=42
)
```

```python
rf2.fit(X_train, y_train)
```

**Justificación**: Optimizado para seleccionar características relevantes y reducir ruido en la predicción.

### Modelo 3: Random Forest con Optimización de Características

En este modelo implementamos una variante de Random Forest enfocada en la optimización y selección de características. Se recomienda que el número de estimadores (`n_estimators`) esté entre 100 y 300 para lograr un buen equilibrio entre rendimiento y tiempo de cómputo; por ello, seleccionamos el valor máximo de 300. 

Para evitar el sobreajuste y mejorar la capacidad de generalización del modelo, limitamos la profundidad máxima (`max_depth`) de los árboles a 13, siguiendo recomendaciones de la literatura para problemas de clasificación.

También ajustamos otros hiperparámetros importantes:
- `min_samples_split` se estableció en 3 para asegurar que cada división interna de los árboles tenga suficientes muestras, ayudando a reducir el sobreajuste.
- `min_samples_leaf` se fijó en 2 para que cada hoja tenga al menos dos muestras, aportando mayor robustez.
- `max_features='sqrt'` indica que, en cada división, el modelo selecciona aleatoriamente solo la raíz cuadrada del número total de características. Esta técnica fomenta la diversidad entre los árboles y es una práctica recomendada en Random Forest para problemas de clasificación.

```python
# Configuration with feature selection
rf3 = RandomForestClassifier(
    n_estimators=300,
    max_depth=13,
    min_samples_split=3,
    min_samples_leaf=2,
    max_features='sqrt',
    class_weight="balanced",
    random_state=42
)
```

```python
rf3.fit(X_train, y_train)
```

**Justificación**: Optimizado para seleccionar características relevantes y reducir ruido en la predicción.
