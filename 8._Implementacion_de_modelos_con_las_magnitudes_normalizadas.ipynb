### 1. Selección del Algoritmo

Hemos seleccionado **Random Forest** como nuestro algoritmo para crear los modelos.

```python
import pandas as pd
from constants import scaled_dir, predictor_vars, target_var
```

### 2. Creacion de los dataframes

```python
X_val_scaled = pd.read_csv(f"{scaled_dir}/X_val.csv")
X_train_scaled = pd.read_csv(f"{scaled_dir}/X_train.csv")
y_val_scaled = pd.read_csv(f"{scaled_dir}/y_val.csv")
y_train_scaled = pd.read_csv(f"{scaled_dir}/y_train.csv")
```

### 4. Configuración de los Modelos
Implementamos tres variantes del Random Forest con diferentes enfoques:

### Modelo 4: Random Forest con Configuración Básica

```python
from sklearn.ensemble import RandomForestClassifier
```

Empezamos con el modelo mas sencillo donde practicamente usamos los valores por defecto de los parámetros

```python
rf4 = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42
)
```

```python
rf4.fit(X_train_scaled, y_train_scaled)
```

### Modelo 5: Random Forest con Optimización de Características

```python
rf5 = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=2,
    class_weight='balanced',
    random_state=42
)
```

```python
rf5.fit(X_train_scaled, y_train_scaled)
```

**Justificación**: Optimizado para seleccionar características relevantes y reducir ruido en la predicción.

### Modelo 6: Random Forest con Optimización de Características

```python
rf6 = RandomForestClassifier(
    n_estimators=150,
    max_depth=15,
    min_samples_split=3,
    min_samples_leaf=2,
    max_features='sqrt',
    class_weight="balanced",
    random_state=42
)
```

```python
rf6.fit(X_train_scaled, y_train_scaled)
```

**Justificación**: Optimizado para seleccionar características relevantes y reducir ruido en la predicción.
