```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
sns.set_theme(color_codes=True)
```

```python
df = pd.read_csv('bank_churn.csv')
df.head()
```

```python
df_encoded = pd.get_dummies(df, columns=['country'])
df_encoded.head()
```

```python
df_encoded['country_France'] = df_encoded['country_France'].replace([True], '1')
df_encoded['country_France'] = df_encoded['country_France'].replace([False], '0')

df_encoded['country_Germany'] = df_encoded['country_Germany'].replace([True], '1')
df_encoded['country_Germany'] = df_encoded['country_Germany'].replace([False], '0')

df_encoded['country_Spain'] = df_encoded['country_Spain'].replace([True], '1')
df_encoded['country_Spain'] = df_encoded['country_Spain'].replace([False], '0')

df_encoded['country_France'] = df_encoded['country_France'].replace([True], '1')
df_encoded['country_France'] = df_encoded['country_France'].replace([False], '0')

df_encoded['gender'] = df_encoded['gender'].replace(['Female'], '0')
df_encoded['gender'] = df_encoded['gender'].replace(['Male'], '1')

df_encoded['country_France'] = pd.to_numeric(df_encoded['country_France'])
df_encoded['country_Germany'] = pd.to_numeric(df_encoded['country_Germany'])
df_encoded['country_Spain'] = pd.to_numeric(df_encoded['country_Spain'])
df_encoded['gender'] = pd.to_numeric(df_encoded['gender'])
```

```python
df_encoded.head()
```

```python
df_encoded.dtypes
```

```python
df_processed = df_encoded.drop('customer_id', axis=1, inplace=False)
```

```python
df_processed.head()
```

```python
from sklearn.model_selection import train_test_split, cross_val_score, cross_val_predict
from sklearn.metrics import classification_report

X = df_processed.drop('churn', axis=1)
y = df_processed['churn']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn import svm
svm1 = svm.SVC(random_state=0)
svm2 = svm.SVC(random_state=0, class_weight='balanced')

svm1.fit(X_train_scaled, y_train)
svm2.fit(X_train_scaled, y_train)

y_pred_svm1 = svm1.predict(X_test_scaled)
y_pred_svm2 = svm2.predict(X_test_scaled)
```

```python
print(classification_report(y_test, y_pred_svm1))
print(classification_report(y_test, y_pred_svm2))
```

### ¿Cómo difieren las métricas de clasificación para ambos modelos?

1. **Precision**:
   - En la clase mayoritaria (0): El modelo balanceado (svm2) tiene mejor precisión (0.93 vs 0.85).
   - En la clase minoritaria (1): El modelo estándar (svm1) tiene mucho mejor precisión (0.88 vs 0.45).

2. **Recall**:
   - En la clase mayoritaria (0): El modelo estándar tiene mejor recall (0.99 vs 0.76).
   - En la clase minoritaria (1): El modelo balanceado tiene significativamente mejor recall (0.76 vs 0.29).

3. **F1-score**:
   - En la clase mayoritaria (0): El modelo estándar es superior (0.91 vs 0.84).
   - En la clase minoritaria (1): El modelo balanceado es superior (0.57 vs 0.44).

4. **Exactitud (Accuracy)**:
   - El modelo estándar tiene mejor exactitud general (0.85 vs 0.76).

### ¿Existe alguna relación entre ciertas métricas al compararlas en ambos modelos?

1. Existe un claro intecambio de sacreificio entre precission y recall para la clase minoritaria (churn):
   - Al balancear las clases de svm2, el recall para churn mejora significativamente (de 0.29 a 0.76).
   - Sin embargo, esto viene con un costo de reducción en la precisión (de 0.88 a 0.45).

2. El balanceo de clases cambia el enfoque del modelo:
   - El modelo estándar prioriza la clase mayoritaria (no-churn) y tiene dificultades para identificar los casos de churn.
   - El modelo balanceado sacrifica exactitud general para detectar mejor los casos de churn, lo que podría ser más valioso en un contexto de negocio donde identificar clientes en riesgo es prioritario.

3. El F1-score de la clase minoritaria mejora con el balanceo, indicando un mejor equilibrio entre precisión y sensibilidad para estos casos importantes.

## Interpretación de los Resultados de XGBoost

### ¿Los resultados son lo que esperaba? ¿Por qué?

Sí, los resultados siguen el patrón esperado para ajuste de clases desbalanceadas:

1. **Modelo estándar (xgb1):**
   - Alta precisión para clase minoritaria (0.68)
   - Recall moderado para clase minoritaria (0.53)
   - Excelente detección de la clase mayoritaria (0.94)
   - Mejor accuracy general (0.85)

2. **Modelo con scale_pos_weight (xgb2):**
   - Menor precisión para clase minoritaria (0.52 vs 0.68)
   - Mayor recall para clase minoritaria (0.65 vs 0.53)
   - Sacrifica recall en clase mayoritaria (0.85 vs 0.94)
   - Accuracy general ligeramente menor (0.81 vs 0.85)

3. **¿Por qué es lo esperado?**
   - El parámetro `scale_pos_weight` modifica las penalizaciones del modelo para favorecer la clase minoritaria
   - Esto produce el típico trade-off: mejora la detección de casos positivos (churn) a costa de generar más falsos positivos
   - El efecto es similar al observado con `class_weight` en SVM, pero más moderado y controlado
   - Para problemas de churn bancario, este balance puede ser preferible ya que el costo de perder clientes sin identificarlos suele ser mayor que el costo de aplicar estrategias de retención innecesarias

El ajuste mediante `scale_pos_weight` resulta en un modelo más equilibrado para el problema de negocio específico, donde detectar clientes en riesgo de abandono es prioritario aunque implique algunas falsas alarmas.
